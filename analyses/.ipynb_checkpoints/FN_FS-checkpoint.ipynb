```python
def read_fasta(file_path):
    sequences = {}
    with open(file_path, "r") as f:
        seq_id, seq_lines = None, []
        for line in f:
            line = line.strip()
            if line.startswith(">"):
                if seq_id is not None:
                    sequences[seq_id] = "".join(seq_lines)
                seq_id, seq_lines = line[1:], []
            else:
                seq_lines.append(line)
        if seq_id is not None:
            sequences[seq_id] = "".join(seq_lines)
    return sequences


def gid(h):
    return h.split()[0].split("|")[-1]


def extract_cds(cds_fasta, gene_df):
    cds = read_fasta(cds_fasta)
    idx = {gid(h): seq for h, seq in cds.items()}
    ids = gene_df.iloc[:, 0].astype(str).str.strip()
    return {g: idx[g] for g in ids if g in idx}
```

```python
GENETIC_CODE = {
    'TTT':'F','TTC':'F','TTA':'L','TTG':'L','TCT':'S','TCC':'S','TCA':'S','TCG':'S',
    'TAT':'Y','TAC':'Y','TAA':'*','TAG':'*','TGT':'C','TGC':'C','TGA':'*','TGG':'W',
    'CTT':'L','CTC':'L','CTA':'L','CTG':'L','CCT':'P','CCC':'P','CCA':'P','CCG':'P',
    'CAT':'H','CAC':'H','CAA':'Q','CAG':'Q','CGT':'R','CGC':'R','CGA':'R','CGG':'R',
    'ATT':'I','ATC':'I','ATA':'I','ATG':'M','ACT':'T','ACC':'T','ACA':'T','ACG':'T',
    'AAT':'N','AAC':'N','AAA':'K','AAG':'K','AGT':'S','AGC':'S','AGA':'R','AGG':'R',
    'GTT':'V','GTC':'V','GTA':'V','GTG':'V','GCT':'A','GCC':'A','GCA':'A','GCG':'G',
    'GAT':'D','GAC':'D','GAA':'E','GAG':'E','GGT':'G','GGC':'G','GGA':'G','GGG':'G'
}

# fix typo if copied into the dictionary above
GENETIC_CODE['GCG'] = 'A'


def aa_of(codon: str) -> str:
    return GENETIC_CODE.get(codon, 'X')


def count_potential_A_to_G(cds_records):
    if isinstance(cds_records, dict):
        seq_iter = cds_records.values()
    else:
        seq_iter = cds_records

    syn_total = 0
    nonsyn_total = 0

    motifs = ['AA', 'TA', 'CA', 'GA']
    motif_counts    = {m: 0 for m in motifs}
    syn_by_motif    = {m: 0 for m in motifs}
    nonsyn_by_motif = {m: 0 for m in motifs}

    for seq in seq_iter:
        if not seq:
            continue
        s = str(seq).upper().replace('U', 'T')
        L = len(s)
        if L % 3 != 0 or L < 9:
            continue

        n_codons = L // 3
        for codon_idx in range(1, n_codons - 1):
            codon_start = codon_idx * 3
            codon = s[codon_start:codon_start + 3]
            orig_aa = aa_of(codon)
            if orig_aa in ('*', 'X'): # skip STOP codons
                continue

            for pos_in_codon in (0, 1, 2):
                i = codon_start + pos_in_codon
                if s[i] != 'A':
                    continue

                prev = s[i - 1] if i > 0 else 'N'
                if prev not in 'ACGT':
                    continue

                mut = list(codon)
                mut[pos_in_codon] = 'G'
                mut_codon = ''.join(mut)
                mut_aa = aa_of(mut_codon)
                if mut_aa == 'X':
                    continue

                motif = prev + 'A'
                if motif in motif_counts:
                    motif_counts[motif] += 1

                if mut_aa == orig_aa:
                    syn_total += 1
                    if motif in syn_by_motif:
                        syn_by_motif[motif] += 1
                else:
                    nonsyn_total += 1
                    if motif in nonsyn_by_motif:
                        nonsyn_by_motif[motif] += 1

    return {
        'syn_total': syn_total,
        'nonsyn_total': nonsyn_total,
        'motif_counts': motif_counts,
        'syn_by_motif': syn_by_motif,
        'nonsyn_by_motif': nonsyn_by_motif,
    }
```

```python
MOTIFS = ("TA", "AA", "CA", "GA")
BASE_TO_MOTIF = {"T": "TA", "A": "AA", "C": "CA", "G": "GA"}


def edited_by_motif_from_dfs(nonsyn_df_sub, syn_df_sub, motifs=MOTIFS):
    x = pd.concat([nonsyn_df_sub["base_before_edit"], syn_df_sub["base_before_edit"]], ignore_index=True)
    x = x.astype(str).str.strip().str.upper().str[0]
    x = x[x.isin(list("ATCG"))].map(BASE_TO_MOTIF)
    vc = x.value_counts().to_dict()
    return {m: int(vc.get(m, 0)) for m in motifs}


def motif_edit_probs(edited_by_motif: dict, motif_counts: dict, motifs=MOTIFS):
    return {
        m: (edited_by_motif.get(m, 0) / motif_counts.get(m, 0)) if motif_counts.get(m, 0) else 0.0
        for m in motifs
    }


def compute_paragraph_quantities(res, edited_by_motif, obs_n, obs_s, motifs=MOTIFS):
    motif_counts = res["motif_counts"]
    P = motif_edit_probs(edited_by_motif, motif_counts, motifs=motifs)

    # Step 2: S and N
    S = int(res["syn_total"])
    N = int(res["nonsyn_total"])
    total_A = S + N

    # Step 3: fmotif-S and fmotif-N
    fS = {m: int(res["syn_by_motif"].get(m, 0)) for m in motifs}
    fN = {m: int(res["nonsyn_by_motif"].get(m, 0)) for m in motifs}

    # Step 4: E(s) and E(n)
    Es = sum(P[m] * fS[m] for m in motifs)
    En = sum(P[m] * fN[m] for m in motifs)

    # Step 5: S' and N'
    denom = Es + En
    Sprime = (total_A * Es / denom) if denom else np.nan
    Nprime = (total_A * En / denom) if denom else np.nan

    # Step 6: FS and FN
    FS = (obs_s / Sprime) if Sprime > 0 else np.nan
    FN = (obs_n / Nprime) if Nprime > 0 else np.nan

    return dict(
        motif_counts=motif_counts,
        edited_by_motif=edited_by_motif,
        P=P,
        S=S, N=N, total_A=total_A,
        fS=fS, fN=fN,
        Es=Es, En=En,
        Sprime=Sprime, Nprime=Nprime,
        FS=FS, FN=FN,
        obs_s=obs_s, obs_n=obs_n,
    )
```

```python
# use sex_df as the main table for Fig 1b, Fig 1c, and Fig 2a
specs = {
    ("FG", "sex"): {
        "nonsyn_df": fg_nonsyn_sex,
        "syn_df": fg_syn_sex,
        "genes": fg_sex_genes,
        "res": fg_results,
    },
    ("NC", "sex"): {
        "nonsyn_df": nc_nonsyn_sex,
        "syn_df": nc_syn_sex,
        "genes": nc_sex_genes,
        "res": nc_results,
    },
    ("NT", "sex"): {
        "nonsyn_df": nt_nonsyn_sex,
        "syn_df": nt_syn_sex,
        "genes": nt_sex_genes,
        "res": nt_results,
    },
}

rows = []
for (sp, cond), d in specs.items():
    nonsyn_sub = d["nonsyn_df"][d["nonsyn_df"]["GeneID(Symbol)"].astype(str).str.strip().isin(d["genes"])]
    syn_sub    = d["syn_df"]   [d["syn_df"]   ["GeneID(Symbol)"].astype(str).str.strip().isin(d["genes"])]

    obs_n, obs_s = int(len(nonsyn_sub)), int(len(syn_sub))
    edited = edited_by_motif_from_dfs(nonsyn_sub, syn_sub)
    q = compute_paragraph_quantities(d["res"], edited, obs_n, obs_s)

    row = {
        "species": sp,
        "condition": cond,
        "P_TA": q["P"]["TA"],
        "P_AA": q["P"]["AA"],
        "P_CA": q["P"]["CA"],
        "P_GA": q["P"]["GA"],
        "S": q["S"],
        "N": q["N"],
        "S+N": q["total_A"],
        "f_TA-S": q["fS"]["TA"], "f_AA-S": q["fS"]["AA"], "f_CA-S": q["fS"]["CA"], "f_GA-S": q["fS"]["GA"],
        "f_TA-N": q["fN"]["TA"], "f_AA-N": q["fN"]["AA"], "f_CA-N": q["fN"]["CA"], "f_GA-N": q["fN"]["GA"],
        "E(s)": q["Es"],
        "E(n)": q["En"],
        "S'": q["Sprime"],
        "N'": q["Nprime"],
        "s": q["obs_s"],
        "n": q["obs_n"],
        "FS": q["FS"],
        "FN": q["FN"],
        "Fs": q["FS"],
        "Fn": q["FN"],
    }
    rows.append(row)

sex_df = pd.DataFrame(rows).sort_values(["species"]).reset_index(drop=True)
sex_df
```

```python
MOTIFS = ("TA", "AA", "CA", "GA")
BASE_TO_MOTIF = {"T": "TA", "A": "AA", "C": "CA", "G": "GA"}


def edited_by_motif_one_df(df, motifs=MOTIFS):
    """
    Count observed edited sites by motif for one dataframe.

    For synonymous df, this gives:
        s_TA, s_AA, s_CA, s_GA

    For nonsynonymous df, this gives:
        n_TA, n_AA, n_CA, n_GA
    """
    x = df["base_before_edit"].astype(str).str.strip().str.upper().str[0]
    x = x[x.isin(list("ATCG"))].map(BASE_TO_MOTIF)

    vc = x.value_counts().to_dict()
    return {m: int(vc.get(m, 0)) for m in motifs}


def edited_by_motif_from_dfs(nonsyn_df_sub, syn_df_sub, motifs=MOTIFS):
    """
    Count pooled observed edited sites by motif.

    e_motif = s_motif + n_motif

    These pooled counts are used to calculate P_motif.
    """
    syn_counts = edited_by_motif_one_df(syn_df_sub, motifs=motifs)
    nonsyn_counts = edited_by_motif_one_df(nonsyn_df_sub, motifs=motifs)

    return {
        m: syn_counts[m] + nonsyn_counts[m]
        for m in motifs
    }


def motif_edit_probs(edited_by_motif: dict, motif_counts: dict, motifs=MOTIFS):
    """
    P_motif = observed edited sites in motif / total editable/background sites in motif
    """
    return {
        m: (
            edited_by_motif.get(m, 0) / motif_counts.get(m, 0)
            if motif_counts.get(m, 0)
            else 0.0
        )
        for m in motifs
    }


def compute_paragraph_quantities(res, edited_by_motif, obs_n, obs_s, motifs=MOTIFS):
    motif_counts = res["motif_counts"]
    P = motif_edit_probs(edited_by_motif, motif_counts, motifs=motifs)

    # Step 2: S and N
    S = int(res["syn_total"])
    N = int(res["nonsyn_total"])
    total_A = S + N

    # Step 3: fmotif-S and fmotif-N
    fS = {m: int(res["syn_by_motif"].get(m, 0)) for m in motifs}
    fN = {m: int(res["nonsyn_by_motif"].get(m, 0)) for m in motifs}

    # Step 4: E(s) and E(n)
    Es = sum(P[m] * fS[m] for m in motifs)
    En = sum(P[m] * fN[m] for m in motifs)

    # Step 5: S' and N'
    denom = Es + En
    Sprime = (total_A * Es / denom) if denom else np.nan
    Nprime = (total_A * En / denom) if denom else np.nan

    # Step 6: FS and FN
    FS = (obs_s / Sprime) if Sprime > 0 else np.nan
    FN = (obs_n / Nprime) if Nprime > 0 else np.nan

    return dict(
        motif_counts=motif_counts,
        edited_by_motif=edited_by_motif,
        P=P,
        S=S,
        N=N,
        total_A=total_A,
        fS=fS,
        fN=fN,
        Es=Es,
        En=En,
        Sprime=Sprime,
        Nprime=Nprime,
        FS=FS,
        FN=FN,
        obs_s=obs_s,
        obs_n=obs_n,
    )


# use sex_df as the main table for Fig 1b, Fig 1c, and Fig 2a
specs = {
    ("FG", "sex"): {
        "nonsyn_df": fg_nonsyn_sex,
        "syn_df": fg_syn_sex,
        "genes": fg_sex_genes,
        "res": fg_results,
    },
    ("NC", "sex"): {
        "nonsyn_df": nc_nonsyn_sex,
        "syn_df": nc_syn_sex,
        "genes": nc_sex_genes,
        "res": nc_results,
    },
    ("NT", "sex"): {
        "nonsyn_df": nt_nonsyn_sex,
        "syn_df": nt_syn_sex,
        "genes": nt_sex_genes,
        "res": nt_results,
    },
}


rows = []

for (sp, cond), d in specs.items():

    nonsyn_sub = d["nonsyn_df"][
        d["nonsyn_df"]["GeneID(Symbol)"]
        .astype(str)
        .str.strip()
        .isin(d["genes"])
    ]

    syn_sub = d["syn_df"][
        d["syn_df"]["GeneID(Symbol)"]
        .astype(str)
        .str.strip()
        .isin(d["genes"])
    ]

    # observed nonsynonymous and synonymous edited sites
    obs_n = int(len(nonsyn_sub))
    obs_s = int(len(syn_sub))

    # separate observed edited-site motif counts
    syn_edited_by_motif = edited_by_motif_one_df(syn_sub)
    nonsyn_edited_by_motif = edited_by_motif_one_df(nonsyn_sub)

    # pooled observed edited-site motif counts used for P_motif
    edited = {
        m: syn_edited_by_motif[m] + nonsyn_edited_by_motif[m]
        for m in MOTIFS
    }

    q = compute_paragraph_quantities(
        d["res"],
        edited,
        obs_n,
        obs_s,
        motifs=MOTIFS,
    )

    row = {
        "species": sp,
        "condition": cond,

        # observed synonymous edited sites by motif
        "s_TA": syn_edited_by_motif["TA"],
        "s_AA": syn_edited_by_motif["AA"],
        "s_CA": syn_edited_by_motif["CA"],
        "s_GA": syn_edited_by_motif["GA"],

        # observed nonsynonymous edited sites by motif
        "n_TA": nonsyn_edited_by_motif["TA"],
        "n_AA": nonsyn_edited_by_motif["AA"],
        "n_CA": nonsyn_edited_by_motif["CA"],
        "n_GA": nonsyn_edited_by_motif["GA"],

        # pooled observed edited sites by motif
        # e_motif = s_motif + n_motif
        "e_TA": edited["TA"],
        "e_AA": edited["AA"],
        "e_CA": edited["CA"],
        "e_GA": edited["GA"],

        # motif-specific editing probabilities
        "P_TA": q["P"]["TA"],
        "P_AA": q["P"]["AA"],
        "P_CA": q["P"]["CA"],
        "P_GA": q["P"]["GA"],

        # total available synonymous / nonsynonymous sites
        "S": q["S"],
        "N": q["N"],
        "S+N": q["total_A"],

        # available synonymous sites by motif
        "f_TA-S": q["fS"]["TA"],
        "f_AA-S": q["fS"]["AA"],
        "f_CA-S": q["fS"]["CA"],
        "f_GA-S": q["fS"]["GA"],

        # available nonsynonymous sites by motif
        "f_TA-N": q["fN"]["TA"],
        "f_AA-N": q["fN"]["AA"],
        "f_CA-N": q["fN"]["CA"],
        "f_GA-N": q["fN"]["GA"],

        # motif-adjusted expected counts
        "E(s)": q["Es"],
        "E(n)": q["En"],
        "S'": q["Sprime"],
        "N'": q["Nprime"],

        # observed total synonymous / nonsynonymous edited sites
        "s": q["obs_s"],
        "n": q["obs_n"],

        # observed / expected ratios
        "FS": q["FS"],
        "FN": q["FN"],
        "Fs": q["FS"],
        "Fn": q["FN"],
    }

    rows.append(row)


sex_df = (
    pd.DataFrame(rows)
    .sort_values(["species"])
    .reset_index(drop=True)
)

sex_df
sex_df.to_csv('/Users/jiachenli/Desktop/new_data.csv',index=False)
```

```python
def p_to_sig(pval):
    if pval < 1e-4:
        return "****"
    elif pval < 1e-3:
        return "***"
    elif pval < 1e-2:
        return "**"
    elif pval < 5e-2:
        return "*"
    else:
        return "ns"


def chi2_from_sex_df(row):
    obs_n = row["n"]
    obs_s = row["s"]
    Np = row["N'"]
    Sp = row["S'"]

    if not np.isfinite([obs_n, obs_s, Np, Sp]).all():
        return np.nan
    if Np <= 0 or Sp <= 0:
        return np.nan

    table = np.array([
        [obs_n, Np - obs_n],
        [obs_s, Sp - obs_s],
    ], dtype=float)

    if (table < 0).any():
        return np.nan

    return chi2_contingency(table).pvalue


sex_df["p"] = sex_df.apply(chi2_from_sex_df, axis=1)

# results_df is kept for Fig 2a, but it comes from sex_df
results_df = sex_df[["species", "condition", "n", "s", "N", "S", "N'", "S'", "Fn", "Fs", "p"]].copy()
results_df
```

```python
# bootstrap samples for Fig 1c
fg_bootstrap_smaples = pd.read_csv('/Users/jiachenli/Desktop/fungal_rna_editing/main/bootstrap/0428_cleaned_sites/fg_bootstrap_1000.csv')
nc_bootstrap_smaples = pd.read_csv('/Users/jiachenli/Desktop/fungal_rna_editing/main/bootstrap/0428_cleaned_sites/nc_bootstrap_1000.csv')
nt_bootstrap_smaples = pd.read_csv('/Users/jiachenli/Desktop/fungal_rna_editing/main/bootstrap/0428_cleaned_sites/nt_bootstrap_1000.csv')

for boot in [fg_bootstrap_smaples, nc_bootstrap_smaples, nt_bootstrap_smaples]:
    boot['N/S'] = boot['N'] / boot['S']
    boot['(N_prime/S_pirme)'] = boot['N_prime'] / boot['S_prime']
    boot['(N_prime/S_pirme / N/S)'] = boot['(N_prime/S_pirme)'] / boot['N/S']
```

```python
# Fig 1c
needed1c = ["species", "N", "S", "N'", "S'"]
missing = [c for c in needed1c if c not in sex_df.columns]
if missing:
    raise ValueError(f"Missing columns for Fig1c: {missing}")

df = sex_df[needed1c].copy()
df["N_over_S"] = np.where(df["S"] != 0, df["N"] / df["S"], np.nan)
df["Np_over_Sp"] = np.where(df["S'"] != 0, df["N'"] / df["S'"], np.nan)

wide_ratio = (
    df.set_index("species")[["N_over_S", "Np_over_Sp"]]
      .rename(index=SPECIES_RENAME)
)


def _bootstrap_ci(samples_1d, lo=2.5, hi=97.5):
    x = np.asarray(samples_1d, dtype=float)
    x = x[np.isfinite(x)]
    if x.size == 0:
        return (np.nan, np.nan)
    return (np.percentile(x, lo), np.percentile(x, hi))


def _centered_bootstrap_yerr(samples_1d, lo=2.5, hi=97.5):
    # use bootstrap spread for the error size, but keep sex_df as the bar height
    x = np.asarray(samples_1d, dtype=float)
    x = x[np.isfinite(x)]
    if x.size == 0:
        return (np.nan, np.nan)

    center = np.median(x)
    dev = x - center
    dev_lo, dev_hi = np.percentile(dev, [lo, hi])
    return max(-dev_lo, 0), max(dev_hi, 0)


boot_by_species = {
    SPECIES_RENAME["FG"]: fg_bootstrap_smaples,
    SPECIES_RENAME["NC"]: nc_bootstrap_smaples,
    SPECIES_RENAME["NT"]: nt_bootstrap_smaples,
}

for sp, boot in boot_by_species.items():
    required_boot_cols = ["N", "S", "N_prime", "S_prime"]
    missing_boot = [c for c in required_boot_cols if c not in boot.columns]
    if missing_boot:
        raise ValueError(f"{sp} bootstrap df missing columns: {missing_boot}")

    boot["N_over_S"] = np.where(boot["S"] != 0, boot["N"] / boot["S"], np.nan)
    boot["Np_over_Sp"] = np.where(boot["S_prime"] != 0, boot["N_prime"] / boot["S_prime"], np.nan)

ci_low = pd.DataFrame(index=wide_ratio.index, columns=wide_ratio.columns, dtype=float)
ci_high = pd.DataFrame(index=wide_ratio.index, columns=wide_ratio.columns, dtype=float)
yerr_lower = pd.DataFrame(index=wide_ratio.index, columns=wide_ratio.columns, dtype=float)
yerr_upper = pd.DataFrame(index=wide_ratio.index, columns=wide_ratio.columns, dtype=float)

for sp in wide_ratio.index:
    boot = boot_by_species[sp]

    for col in wide_ratio.columns:
        lo, hi = _bootstrap_ci(boot[col])
        ci_low.loc[sp, col] = lo
        ci_high.loc[sp, col] = hi

        lower, upper = _centered_bootstrap_yerr(boot[col])
        yerr_lower.loc[sp, col] = lower
        yerr_upper.loc[sp, col] = upper

# raw percentile CI can be shifted slightly from the observed estimate for ratios
outside_raw_ci = (wide_ratio < ci_low) | (wide_ratio > ci_high)
if outside_raw_ci.any().any():
    print("Note: raw percentile CI does not contain the point estimate for some ratios.")

yerr = np.stack([yerr_lower.values.T, yerr_upper.values.T], axis=1)

fig1c, ax1c = plt.subplots(figsize=(13, 6))

wide_ratio.plot(
    kind="bar",
    ax=ax1c,
    color=["white", "dimgrey"],
    yerr=yerr,
    capsize=4,
    legend=False,
    error_kw=dict(elinewidth=1.5, capthick=1.5),
    edgecolor="black",
    linewidth=1.25,
)

ax1c.tick_params(axis="x", labelsize=24, rotation=0)
ax1c.tick_params(axis="y", labelsize=24)
ax1c.set_xlabel("")
ax1c.set_ylabel(r"$\it{N}/\it{S}$ and $\it{N}'/\it{S}'$", fontsize=24)

sns.despine(top=True, right=True)
fig1c.subplots_adjust(right=0.82)

ax1c.legend(
    [r"$\it{N}/\it{S}$", r"$\it{N}'/\it{S}'$"],
    title="",
    frameon=False,
    fontsize=23,
    title_fontsize=24,
    loc="upper left",
)

orange = ax1c.containers[1]

ymax = np.nanmax((wide_ratio + yerr_upper).values)
pad_default = 0.03 * ymax
pad_first = 0.012 * ymax

for i, bar in enumerate(orange):
    sp = wide_ratio.index[i]
    top = wide_ratio.loc[sp, "Np_over_Sp"] + yerr_upper.loc[sp, "Np_over_Sp"]
    if not np.isfinite(top):
        continue

    x = bar.get_x() + bar.get_width() / 2.0
    pad = pad_first if i == 0 else pad_default

    ax1c.text(
        x,
        top + pad,
        "***",
        ha="center",
        va="bottom",
        fontsize=24,
    )

fig1c.tight_layout()
#fig1c.savefig(f"{fig_dir}/Fig1c_04292026.pdf", dpi=300, bbox_inches="tight")
plt.show()
```

```python
def plot_fnfs_from_results(results_df,
                           out_pdf=None,
                           sp_order=("FG", "NC", "NT"),
                           bar_width=0.45,
                           top_pad_frac=0.01,
                           legend_y=-0.10,
                           figsize=(13, 6),
                           stars_fontsize=22,
                           label_fontsize=22,
                           tick_fontsize=22):

    # use genes that are only expressed during sex
    df = results_df.copy()
    df = df[df["condition"] == "sex"].copy()
    df["species"] = pd.Categorical(df["species"], categories=list(sp_order), ordered=True)
    df = df.sort_values("species").reset_index(drop=True)

    species_meta = [
        ("FG", r"$\it{F.\ graminearum}$"),
        ("NC", r"$\it{N.\ crassa}$"),
        ("NT", r"$\it{N.\ tetrasperma}$"),
    ]

    species_data = []
    for sp_key, display_name in species_meta:
        sub = df[df["species"] == sp_key]
        if sub.empty:
            species_data.append({"key": sp_key, "name": display_name, "Fn": np.nan, "Fs": np.nan, "p": np.nan})
            continue
        species_data.append({
            "key": sp_key,
            "name": display_name,
            "Fn": float(sub["Fn"].iloc[0]),
            "Fs": float(sub["Fs"].iloc[0]),
            "p":  float(sub["p"].iloc[0]),
        })

    fig, axes = plt.subplots(1, len(species_data), figsize=figsize, sharey=True)
    if len(species_data) == 1:
        axes = [axes]

    finite_vals = []
    for d in species_data:
        if np.isfinite(d["Fn"]): finite_vals.append(d["Fn"])
        if np.isfinite(d["Fs"]): finite_vals.append(d["Fs"])
    ymax = max(finite_vals) if finite_vals else 0.0
    y_pad = top_pad_frac * ymax if ymax > 0 else 1e-4

    for i, (ax, d) in enumerate(zip(axes, species_data)):
        x = 0.5
        x_syn = x - bar_width / 2
        x_nonsyn = x + bar_width / 2

        if not (np.isfinite(d["Fn"]) and np.isfinite(d["Fs"])):
            ax.axis("off")
            continue

        ax.bar(x_syn, d["Fs"], width=bar_width, color="grey", edgecolor="none")
        ax.bar(x_nonsyn, d["Fn"], width=bar_width, color="black", edgecolor="none")

        sig = p_to_sig(d["p"]) if np.isfinite(d["p"]) else "ns"
        ax.text(x_nonsyn, d["Fn"] + y_pad, sig, ha="center", va="bottom", fontsize=stars_fontsize)

        ax.set_xticks([x])
        ax.set_xticklabels([d["name"]], fontsize=tick_fontsize)
        ax.set_xlim(0, 1)

        if i == 0:
            ax.set_ylabel("Fraction of sites edited", fontsize=label_fontsize)
            ax.tick_params(axis="y", labelsize=tick_fontsize)
        else:
            ax.tick_params(axis="y", left=False, labelleft=False)
            ax.set_ylabel("")

        sns.despine(ax=ax, top=True, right=True, left=(i != 0))

    handles = [
        plt.Rectangle((0, 0), 1, 1, color="grey",  label=r"Synonymous"),
        plt.Rectangle((0, 0), 1, 1, color="black", label=r"Nonsynonymous")
    ]
    fig.legend(handles=handles, fontsize=label_fontsize, frameon=False,
               loc="lower center", bbox_to_anchor=(0.5, legend_y), ncol=2)

    plt.tight_layout(rect=[0.03, 0.05, 1, 0.95])

    if out_pdf is not None:
        fig.savefig(out_pdf, dpi=300, bbox_inches="tight", pad_inches=0.2)

    plt.show()
    return fig, axes


plot_fnfs_from_results(results_df, out_pdf=f'{fig_dir}/Fig2a_04292026.pdf')
```

```python
def plot_syn_nonsyn_boxplot(species_dict, ylabel="Editing Level (%)", title=None, save_path=None):
    box_colors = ['grey', 'black']

    species_labels = {
        'fg': r"$\it{F.\ graminearum}$",
        'nc': r"$\it{N.\ crassa}$",
        'nt': r"$\it{N.\ tetrasperma}$"
    }
    species_order = list(species_dict.keys())

    all_data = []
    for sp in species_order:
        s_edit = species_dict[sp].get("syn", pd.Series(dtype=float)).dropna()
        n_edit = species_dict[sp].get("nonsyn", pd.Series(dtype=float)).dropna()
        all_data.append([s_edit, n_edit])

    fig, ax = plt.subplots(figsize=(13, 6))
    positions, box_data, colors, xticks, xtick_labels = [], [], [], [], []

    spacing, width = 3, 0.6

    for i, (sp, (s, n)) in enumerate(zip(species_order, all_data)):
        base = i * spacing
        box_data.extend([s, n])
        positions.extend([base, base + 1])
        colors.extend(box_colors)
        xticks.append(base + 0.5)
        xtick_labels.append(species_labels.get(sp, sp))

    bp = ax.boxplot(
        box_data,
        positions=positions,
        widths=width,
        patch_artist=True,
        medianprops=dict(color='white', linewidth=1.5),
        showfliers=False
    )

    for patch, color in zip(bp['boxes'], colors):
        patch.set_facecolor(color)
        patch.set_linewidth(0)

    ax.set_ylabel(ylabel, fontsize=22)
    ax.set_xticks(xticks)
    ax.set_xticklabels(xtick_labels, fontsize=22)
    ax.tick_params(axis='y', labelsize=22)

    ax.set_ylim(-2, 80)
    ymin, ytop = ax.get_ylim()
    yrange = ytop - ymin

    if title:
        ax.set_title(title, fontsize=22)

    for i, (s, n) in enumerate(all_data):
        base = i * spacing
        if len(s) == 0 or len(n) == 0:
            continue

        y_max = np.nanmax([s.max(), n.max()])

        bar_gap = 0.13 * yrange
        text_gap = 0.12 * yrange

        line_y = min(y_max + bar_gap, ytop - 0.10 * yrange)
        text_y = min(line_y + text_gap, ytop - 0.04 * yrange)

        try:
            pval = mannwhitneyu(n, s, alternative='two-sided').pvalue
            stars = p_to_sig(pval)

            ax.plot([base, base + 1], [line_y, line_y], color='black', linewidth=1)
            ax.text(base + 0.5, text_y, stars, ha='center', fontsize=22)
        except Exception:
            pass

    sns.despine()
    plt.tight_layout(rect=[0, 0.05, 1, 1])

    if save_path:
        fig.savefig(save_path, dpi=300, bbox_inches='tight')

    plt.show()
    return fig, ax
```
